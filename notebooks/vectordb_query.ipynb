```python
import chromadb
import os

# 1. Conectarse a la base de datos (reemplaza 'ruta/a/tus/datos' por tu directorio)
# cliente = chromadb.PersistentClient(path='/Users/gblasd/Documents/SmartBnB/db/chroma_db')
cliente = chromadb.PersistentClient(path=os.path.abspath(os.path.join("..", "db", "chroma_db")))

# 2. Obtener la colección (reemplaza 'nombre_coleccion' por el nombre de tu colección)
coleccion = cliente.get_collection(name="smartbnb_vector_store")

# 3. Obtener todos los datos
# Incluye "documents", "metadatas", "ids" y opcionalmente "embeddings"
todos_los_datos = coleccion.get(
    #include=["documents", "metadatas"]
)

# 4. Imprimir o procesar los datos
print("Total de registros:", len(todos_los_datos["ids"]))

for idx, doc, meta in zip(
    todos_los_datos["ids"], 
    todos_los_datos["documents"], 
    todos_los_datos["metadatas"]
):
    print(f"ID: {idx}")
    print(f"Documento: {doc}")
    print(f"Metadatos: {meta}")
    print("-" * 40)

    
```
